Four principles of OOP

1. Encapsulation

    Encapsulation is used to protect data, control
    access, validate data, be easy to maintenance, reduce error, and improve security.

```python
class Car:
    def __init__(self):
        self.__max_speed = 200

    def get_max_speed(self):
        return self.__max_speed

car = Car()
print(car.get_max_speed())
```

2. Abstraction

    Abstraction is used to hide complicated code, make programs easier to use, reduce complexity, and show only necessary information to the user.

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

3. Inheritance

    Inheritance is used to allow a child class to reciev attributes and functions from a parent class. It helps us reuse code and reduce code duplication.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

dog = Dog()
dog.speak()
```

4. Polymorphism

    Polymorphism is used to allow the same method name to perfom different actions depending on the object.

```python
class Dog:
    def speak(aelf):
        print("Dog barks ")

class Cat:
    def speak(self):
        print("Cat meows")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

Homework

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price 
    
book = Book("Python for Data Analysis", "Wes McKinney", 25)
print("\n===== Encapsulation =====")
print("Title: ", book.title)
print("Author: ", book.author)
print("Price: ", book.get_price())
```

```python
class EBook(Book):
    def __init__(self, title, author, price, quantity):
        super().__init__(title, author, price)
        self.quantity = quantity

ebook = EBook("Introduction to Programming using JAVA", "David J. Eck", 55, 10)
print("\n===== Inheritance =====")
print("Title: ", ebook.title)
print("Author: ", ebook.author)
print("Price: ", ebook.get_price())
print("Quantity: ", ebook.quantity)
```
